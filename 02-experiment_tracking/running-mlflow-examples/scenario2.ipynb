Mlflow setup:

    * tracking server: yes, a local server
    * backend store: sqlite database
    * artifacts store: local filesystem

The experiments can be explored locally by accessing the local tracking server
To run this example you need to launch the mlflow server locally by running the following command in your terminal:

<span style="color: yellow">mlflow server --backend-store-uri sqlite:///backend.db --default-artifact-root ./artifacts_local</span>
#### not 'mlflow ui' but the actual 'server'

```python
import mlflow 
mlflow.set_tracking_uri("http://127.0.0.1:5000") # this is the address provided after running 'mlflow server' command above
```

```python
# double check the tracking uri
print(f"tracking URI: {mlflow.get_tracking_uri()}")
```

```python
from mlflow.tracking import MlflowClient
client = MlflowClient()
experiments = client.search_experiments()
for exp in experiments:
    print(f"name={exp.name}, experiment_id={exp.experiment_id}, artifact_location={exp.artifact_location}, lifecycle_stage={exp.lifecycle_stage}, tag={exp.tags}") 
    # whenever you start using mlflow there will be a default experiment with id = 0
    # artifacts_local folder was created automatically
    # backend.db was created automatically
```
